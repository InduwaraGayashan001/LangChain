# The Stateless Problem

````python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model = "openai/gpt-4o-mini",
    api_key=userdata.get('GITHUB_TOKEN'),
    base_url="https://models.github.ai/inference"
)

result1 = llm.invoke("Who won 1996 Cricket World Cup?")
print(f"Answer for Q1: {result1.content}")
result2 = llm.invoke("What is the largest country?")
print(f"Answer for Q2: {result2.content}")
result3 = llm.invoke("Who was the winning captain of that team?")
print(f"Answer for Q3: {result3.content}")
````

# Memory

## ConversationBufferMemory

````python
from langchain_openai import ChatOpenAI
from langchain_classic.chains import ConversationChain
from langchain_classic.memory import ConversationBufferMemory

llm = ChatOpenAI(
    model = "openai/gpt-4o-mini",
    api_key=userdata.get('GITHUB_TOKEN'),
    base_url="https://models.github.ai/inference"
)

memory = ConversationBufferMemory()

convo = ConversationChain(llm=llm, verbose=True,memory=memory)

convo.invoke("Who won 1996 Criket World Cup?")
````

````python
convo.invoke("What is the largest country?")
````

````python
convo.invoke("Who was the winning captain of that team?")
````

````python
print(convo.memory.buffer)
````

## ConversationSummaryMeomory

````python
from langchain_openai import ChatOpenAI
from langchain_classic.chains import ConversationChain
from langchain_classic.memory import ConversationSummaryMemory

llm = ChatOpenAI(
    model = "openai/gpt-4o-mini",
    api_key=userdata.get('GITHUB_TOKEN'),
    base_url="https://models.github.ai/inference"
)

summary_memory = ConversationSummaryMemory(llm=llm)

summary_convo = ConversationChain(llm=llm, verbose=True, memory=summary_memory)
summary_convo.invoke("Who won 1996 Criket World Cup?")
````

````python
summary_convo.invoke("What is the largest country?")
````

````python
summary_convo.invoke("Who was the winning captain of that team?")
````

````python
print(summary_convo.memory.buffer)
````

## ConversationBufferWindowMeomory

````python
from langchain_openai import ChatOpenAI
from langchain_classic.chains import ConversationChain
from langchain_classic.memory import ConversationBufferWindowMemory

llm = ChatOpenAI(
    model = "openai/gpt-4o-mini",
    api_key=userdata.get('GITHUB_TOKEN'),
    base_url="https://models.github.ai/inference"
)

window_memory = ConversationBufferWindowMemory(k=2)

window_convo = ConversationChain(llm=llm, verbose=True, memory=window_memory)

window_convo.invoke("Who won 1996 Criket World Cup?")
````

````python
window_convo.invoke("What is the largest country?")
````

````python
window_convo.invoke("What is the smallest contry?")
````

````python
window_convo.invoke("Who was the winning captain?")
````

````python
print(window_convo.memory.buffer)
````

## ConversationSummaryBufferMeomory

````python
from langchain_classic.memory import ConversationSummaryBufferMemory
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint

llama_endpoint = HuggingFaceEndpoint(
  repo_id="meta-llama/Llama-3.1-8B-Instruct",
  temperature=0.0,
  huggingfacehub_api_token =userdata.get('HF_TOKEN')
)
llama_llm = ChatHuggingFace(llm=llama_endpoint)

summary_buffer_memory = ConversationSummaryBufferMemory(llm=llama_llm, max_token_limit=100)
summary_buffer_convo = ConversationChain(llm=llama_llm, verbose=True, memory=summary_buffer_memory)
summary_buffer_convo.invoke("Who won 1996 Criket World Cup?")
````

````python
summary_buffer_convo.invoke("What is the largest country?")
````

````python
summary_buffer_convo.invoke("Who was the winning captain?")
````

````python
print(summary_buffer_convo.memory.buffer)
````

````python
summary_buffer_memory = ConversationSummaryBufferMemory(llm=llama_llm, max_token_limit=1000)
summary_buffer_convo = ConversationChain(llm=llama_llm, memory=summary_buffer_memory)
summary_buffer_convo.invoke("Who won 1996 Criket World Cup?")
summary_buffer_convo.invoke("What is the largest country?")
summary_buffer_convo.invoke("Who was the winning captain?")
print(summary_buffer_convo.memory.buffer)
````
