#Calling a Llama Model via Hugging Face

```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint

llama_endpoint = HuggingFaceEndpoint(
  repo_id="meta-llama/Llama-3.1-8B-Instruct",
  temperature=0.1,
  max_new_tokens=2,
  huggingfacehub_api_token =userdata.get('HF_TOKEN')
)
llm = ChatHuggingFace(llm=llama_endpoint)
result = llm.invoke("What is the capital of India? Just give me the name of the city.")
print(result.content)
```

#Calling OpenAI Models via GitHub Marketplace

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model="openai/gpt-4o-mini",
    api_key=userdata.get('GITHUB_TOKEN'),
    base_url="https://models.github.ai/inference"
)
result = llm.invoke(
    "What is the capital of India? Just give me the name of the city."
)
print(result.content)
```
